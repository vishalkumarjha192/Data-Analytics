```python
# Type casting
"""There are two types of types-casting:
1. Implicit type conversion: where python itself convert one datatype to another.
2. Explicit Type Conversion: where the user converts on datatype to another.
"""

name="vishal"
print(type(name))

age=23
print(type(age))

number=25.3
print(type(number))


# implicit type conversion

a=20
b=26.8
print(type(a))
print(type(b))

c=a+b
print(c)
print(type(c))


# Explicit Type Conversion
x="123"
y=10

# z=x+y #give error

z=int(x)+y
print(z)
```
