Types of operators:



Operators can be further divided into 6 categoties:

1. Arithmetic Operators
2. Comparison Operators
3. Logical Operators
4. Assignment Operators
5. Identity Operators
6. Membership Operators
7. Bitwise Operators

```python
# 1. Arithmetic Operators
# Addition (+)
a=36
b=2
print(a+b) #38


# Substraction (-)
print(a-b) #34


# Multiplication (*)
print(a*b) #72

# Division (/)
print(a/b)  #18


# Floor division (//)
print(a//b)  #34


# Modulus (%)
print(a%b) #0

# Exponentiation(**)
print(a**b)  #1296

```

```python
# 2. Comparison Operators
"""
< (less then)
== (equal to)
<= (Less than or equal to)
>= (Greater than or equal to)
!= (Not equal to)

"""

x=87
y=98
z=75

print(x<y)
print(x==y)
print(x<=y)
print(x>=y)
print(x!=y)
```
