Types of operators:



Operators can be further divided into 6 categoties:

1. Arithmetic Operators
2. Comparison Operators
3. Logical Operators
4. Assignment Operators
5. Identity Operators
6. Membership Operators
7. Bitwise Operators

```python
# 1. Arithmetic Operators
# Addition (+)
a=36
b=2
print(a+b) #38


# Substraction (-)
print(a-b) #34


# Multiplication (*)
print(a*b) #72

# Division (/)
print(a/b)  #18


# Floor division (//)
print(a//b)  #34


# Modulus (%)
print(a%b) #0

# Exponentiation(**)
print(a**b)  #1296

```

```python
# 2. Comparison Operators
"""
< (less then)
== (equal to)
<= (Less than or equal to)
>= (Greater than or equal to)
!= (Not equal to)

"""

x=87
y=98
z=75

print(x<y)
print(x==y)
print(x<=y)
print(x>=y)
print(x!=y)
```

```python
# Logical Operators
a=10
b=35
c=5

# and = true if bothe the operation are true
print(a<b and b>c) #True
print(a<b and b<c) #False

#  or= true if either of the operands is true
print(a<b or b<c) #True

# not = true if openrand is false(complements the operand)
print(not(a<b and b>c)) #False


```

```python
# Assignment operators
"""
=> Assignment operators are used in python to assign values to variables.
=> a=6 is a simple assignment operator that assign the value 6 on the right to the variable a on the left.
"""

a= 10
print(a)

a += 30
print(a)

a -= 10
print(a)


a *= 2
print(a)


```
